```python
# Check your actual subject columns
print(df.columns.tolist())

# Use the subjects in your dataset
subjects = ["Python", "Maths", "English", "Database"]

# Calculate total and average marks
df["Total"] = df[subjects].sum(axis=1)
df["Average"] = df[subjects].mean(axis=1).round(2)

# Display the results
display(df[["Name", "Total", "Average"]])
```

```python
# Check the actual column names
print("Columns:", df.columns.tolist())

# Automatically select the subject columns
subjects = ["Python", "Maths", "English", "Database"]

# Keep only subject columns that exist in the dataset
subjects = [s for s in subjects if s in df.columns]

print("Subjects found:", subjects)

# Recalculate total and average
df["Total"] = df[subjects].sum(axis=1)
df["Average"] = df[subjects].mean(axis=1).round(2)

# Assign grades
def assign_grade(average):
    if average >= 90:
        return "A"
    elif average >= 80:
        return "B"
    elif average >= 70:
        return "C"
    elif average >= 60:
        return "D"
    elif average >= 50:
        return "E"
    else:
        return "F"

# Check pass/fail
def check_result(row):
    for subject in subjects:
        if row[subject] < 35:
            return "Fail"
    return "Pass"

df["Grade"] = df["Average"].apply(assign_grade)
df["Result"] = df.apply(check_result, axis=1)

display(df)
```

```python
# Overall class performance
print("Total students:", len(df))

print("Class average:",
      round(df["Average"].mean(), 2))

print("Highest student average:",
      round(df["Average"].max(), 2))

print("Lowest student average:",
      round(df["Average"].min(), 2))

print("Students passed:",
      (df["Result"] == "Pass").sum())

print("Students failed:",
      (df["Result"] == "Fail").sum())
```

```python
# Calculate subject-wise statistics
subject_summary = df[subjects].agg(
    ["mean", "max", "min"]
).round(2)

print("Subject-wise Performance Analysis")
display(subject_summary)

# Find the best-performing subject
subject_averages = df[subjects].mean()

print("Best-performing subject:",
      subject_averages.idxmax())

print("Lowest-performing subject:",
      subject_averages.idxmin())
```

```python
# Display the top 5 students
top_students = df.nlargest(5, "Average")

print("Top 5 Performing Students")
display(
    top_students[
        ["Name", "Total", "Average", "Grade"]
    ]
)
```
